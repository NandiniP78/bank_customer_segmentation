```python
df2 = df.copy()
```

```python
df2.shape
```

```python
# replace NaN values in CustAccountBalance with 0
df2 = df2.fillna({
    'CustAccountBalance': 0
})
```

```python
df2.dropna(inplace=True)
```

```python
df2.isnull().sum()
```

```python
df2[df2["CustomerDOB"] == "1/1/1800"] 
```

```python
df2.drop(df2[df2['CustomerDOB'] == '1/1/1800'].index, inplace=True)
```

```python
df2.shape
```

```python
df2 = df2.reset_index(drop=True)
```

```python
# Infer each row individually and convert unparseable values to NaT
df2['CustomerDOB'] = pd.to_datetime(
    df2['CustomerDOB'], 
    format='mixed', 
    dayfirst=True, 
    errors='coerce'
)
```

```python
# Infer each row individually and convert unparseable values to NaT
df2['TransactionDate'] = pd.to_datetime(
    df2['TransactionDate'], 
    format='mixed', 
    dayfirst=True, 
    errors='coerce'
)
```

```python
# Convert to string and pad to 6 digits (e.g., 91530 -> '091530')
time_str = df2['TransactionTime'].astype(str).str.zfill(6)

# Insert colons -> '23:45:30'
df2['TransactionTime'] = time_str.str.replace(r'(\d{2})(\d{2})(\d{2})', r'\1:\2:\3', regex=True)
```

```python
# Convert to datetime first, then extract the time component
df2['TransactionTime'] = pd.to_datetime(df2['TransactionTime'], format='%H:%M:%S', errors='coerce').dt.time
```

```python
df2.drop(columns='TransactionYear',inplace=True)
```
